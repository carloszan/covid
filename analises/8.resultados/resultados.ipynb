```python
import matplotlib.pyplot as plt
import seaborn as sns

def plot_comparacao_outliers(df1, df2, label1, label2, file_name):
  plt.figure(figsize=(12,5))
  plt.subplot(2,1,1)
  plt.title(label1)
  sns.lineplot(data=df1, y="casosNovos", x='data')
  plt.xticks([])
  plt.xlabel("")

  plt.subplot(2,1,2)
  plt.title(label2)
  sns.lineplot(data=df2, y="novos_casos_novos", x='data')
  plt.xticks([])
  plt.xlabel("")

  plt.savefig(f"{file_name}.png")
```

```python
import pandas as pd

raw_df = pd.read_parquet('0.raw.parquet')
```

```python
gold_df = pd.read_parquet('2.atributos.parquet')
```

```python
city, state = 'Conselheiro Lafaiete', 'MG'

start_date = '2020-03-27'
end_date = '2021-03-27'

raw_city_df = raw_df.sort_values(by=['data'])
raw_city_df = raw_city_df[(raw_city_df['municipio'] == city) & (raw_city_df['estado'] == state) &  (raw_city_df['data'] > start_date) &(raw_city_df['data'] < end_date)]
raw_city_df = raw_city_df.reset_index()

gold_city_df = gold_df.sort_values(by=['data'])
gold_city_df = gold_city_df[(gold_city_df['municipio'] == city) & (gold_city_df['estado'] == state) & (gold_city_df['data'] > start_date) & (gold_city_df['data'] < end_date)]
gold_city_df = gold_city_df.reset_index()

plot_comparacao_outliers(raw_city_df, gold_city_df, f"{city} de 27/03/2020 até 27/03/2021 - Sem limpeza", f"{city} de 27/03/2020 até 27/03/2021 - Com limpeza", 'linha_do_tempo_lafaiete')
```
