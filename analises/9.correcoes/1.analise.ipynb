```python
import pandas as pd
import matplotlib.pyplot as plt

# Suponha que você tenha um DataFrame chamado 'df' com uma coluna de 'valor'
# Exemplo: df = pd.read_csv('seus_dados.csv')
# df['data'] = pd.to_datetime(df['data'])
# df = df.set_index('data')

# Cria um DataFrame de exemplo
data = {'valor': [10, 12, 15, 13, 16, 18, 20, 19, 22, 25, 24, -27, 10, 12, 32]}
df = pd.DataFrame(data)

# Calcula a média móvel de 3 períodos
df['media_movel'] = df['valor'].rolling(window=2).mean()

# Exibe o DataFrame com a nova coluna
print(df)

# Opcional: plota o gráfico
plt.figure(figsize=(10, 6))
plt.plot(df.index, df['valor'], label='Valor Original')
plt.plot(df.index, df['media_movel'], label='Média Móvel (3 períodos)', color='red')
plt.legend()
plt.title('Média Móvel de Séries Temporais')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Suponha que você tenha um DataFrame chamado 'df' com uma coluna de 'valor'
# Exemplo: df = pd.read_csv('seus_dados.csv')
# df['data'] = pd.to_datetime(df['data'])
# df = df.set_index('data')

# Cria um DataFrame de exemplo
data = {'valor': [10, 12, 15, 13, 16, 18, 20, 19, 22, 25, 24, 27, 29, 30, 32]}
df = pd.DataFrame(data)

# Calcula a média móvel de 3 períodos
df['media_movel'] = df['valor'].rolling(window=7).mean()

# Exibe o DataFrame com a nova coluna
print(df)

# Opcional: plota o gráfico
plt.figure(figsize=(10, 6))
plt.plot(df.index, df['valor'], label='Valor Original')
plt.plot(df.index, df['media_movel'], label='Média Móvel (3 períodos)', color='red')
plt.legend()
plt.title('Média Móvel de Séries Temporais')
plt.show()
```
